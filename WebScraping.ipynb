```python
density_soup = BeautifulSoup(density_response.text,"html.parser")
```

```python
type(density_soup)
```

```python
densitytables = density_soup.find_all("table")
```

```python
len(densitytables)
```

```python
densityheaders=densitytables[0].find_all("th")
```

```python
densityheaders
```

```python
densityheaders_tag=densitytables[1].find_all("th")
```

```python
densityheaders_tag
```

```python
densityheaders=[i.text for i in densityheaders_tag]
print(densityheaders)
```

```python
densityrows_tag=densitytables[1].find_all("td")
```

```python
densityrows_tag
```

```python
densityrows_tag=[i.text for i in densityrows_tag]
print(densityrows_tag)
```

```python
density_dict={}
for i in densityheaders:
    print(i)
```

```python
density_dict={}
for i in densityrows_tag:
    print(i)
```

```python
density_dict={}
n=0
for i in densityheaders:
    density_dict[i]=[densityrows_tag[j] for j in range(n,len(densityrows_tag),len(densityheaders))]
    n+=1
density_dict
```

```python
density_df=pd.DataFrame(density_dict)
```

```python
density_df
```

```python
density_df.index=range(1,len(density_df)+1)
density_df
```

```python
density_df.to_csv("density_data.csv")
```

```python
density_df["Population(2021)"] = density_df["Population (2021)"].str.replace(",","").astype(int)
```

```python
density_df["Area km2"] = density_df["Area km2"].str.replace(",","").astype(int)
```

```python
density_df["Population_density"]=density_df["Population(2021)"]/density_df["Area km2"]
```

```python

import matplotlib.pyplot as plt

x = density_df["Country"][:5]
y = density_df["Population_density"][:5]

plt.bar(x, y)
plt.xlabel("Country")
plt.ylabel("Population Density")
plt.title("Top 5 Population Densities")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Plot 1: Bar chart for top 5 countries by Population Density
plt.figure(figsize=(10, 5))
x = density_df["Country"][:5]
y = density_df["Population_density"][:5]
plt.bar(x, y, color="blue")
plt.xlabel("Country")
plt.ylabel("Population Density")
plt.title("Top 5 Countries by Population Density")
plt.xticks(rotation=45)
plt.show()

# Plot 2: Histogram of Population Density distribution
plt.figure(figsize=(8, 5))
plt.hist(density_df["Population_density"], bins=20, color="green", edgecolor="black")
plt.xlabel("Population Density")
plt.ylabel("Frequency")
plt.title("Distribution of Population Density Across Countries")
plt.show()

# Plot 3: Scatter plot of Population vs Area
plt.figure(figsize=(8, 5))
plt.scatter(density_df["Area km2"], density_df["Population(2021)"], color="red", alpha=0.5)
plt.xlabel("Area (km²)")
plt.ylabel("Population (2021)")
plt.title("Population vs Area")
plt.xscale("log")
plt.yscale("log")
plt.grid(True, which="both", linestyle="--", linewidth=0.5)
plt.show()



```
